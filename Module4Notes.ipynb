```python
groceries = {
    "potato" : 3.00,
    "apple" : 1.00,
    "banana" : 0.50,
    "orange" : 1.50,
    "watermelon" : 5.00
}

groceries["strawberries"] = 0.2
print(groceries)
print(groceries["apple"])

#ask user for new fruit and price
#store them as new key/value pair in dictionary

x = input("new fruit")
y = input("new fruit's value")
groceries[x] = y
print(groceries)
```

```python
family = {
    "Julian" : "pepperoni",
    "Adrianna" : "mushroom",
    "Tristan" : "anchovies",
    "Joaquin" : "pineapple",
    "Lucas" : "cheese",
    "Jose" : "pepper",
    "Nola" : "cheese",
}

family["Uncle Mel"] = "pumpernickel"
print(family["Lucas"])

family["Lucas"] = "N/A"
print(f"{family}\n{len(family)}")

print(family["Julian"][1].title())
family["Julian"] = family["Julian"].title()
print(family)

print(family.get("Jose", "Jose not in family"))

for k in family.keys():
    print(f"KEY: {k}")

for v in family.values():
    print(f"VALUE: {v}")
```

```python
census = {
    "household 1": {
        "name":"Blake",
        "age":100,
        "income": 100000000
    },
    "household 2": {
        "name": "Julien",
        "age": 200,
        "income" : [6.5,1,2.6,5.5]
    }
}

print(census["household 2"]["income"][2])
print(len(census), len(census["household 2"]), len(census["household 2"]["income"]))
```

#### lists allow us to store multiple in one place and we access them via index number in the list, dictionaries allow us to store things by key and value, we can access the value by referencing the key. it's important to keep data type we are using in mind
#### above, census is a dictionary, census["household 2"] is also a dictionary, census["household 2"]["income"] is a list, census["household 2"]["income"][2] is an integer
