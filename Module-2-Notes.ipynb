### Pop requires me to know the index of the value I'm trying to remove whereas remove requires me to know the EXACT value.

```python
numbers = [3, 5, 8, 10, 11, 28, 42, 77, 101, 108, 2002]
#Without modifying code above, change the 8 to a 1.
numbers[2] = 1

#Insert a 50 at index 2.
numbers.insert(2, 50)

#Remove the 108 by value.
numbers.remove(108)

#Add 100 to the end.
numbers.append(100)

#Remove 2002 by negative index.
del numbers[-2]

#Remove 11 by positive index.
del numbers[5]

#Use the del function to delete the 5.
del numbers[1]

print(numbers)
```

```python
#Create a list with 5 items then print the list.
listVar = ["One item", "two item", "three item", "four item", "five item"]
print(listVar)
#Append "Hello" to the list and print the list.
listVar.append("Hello")
print(listVar)
#Insert the number 1 to index location 3, then print the list.
listVar.insert(3, 1)
print(listVar)
#Delete the second item then print the list.
del listVar[1]
print(listVar)
#Remove "Hello" then print the list.
listVar.remove("Hello")
print(listVar)
#Pop the last item and print it in a sentence.
pop_var = listVar.pop(4)
print(pop_var)
```

### Pop has three functionalities: it can remove by index (i.e. my_list.pop(1)), it allows you to store the value it deletes in a variable (i.e. y = my_list.pop(1)), it can delete the last item if you don't give it an additional value (i.e. my_list.pop)
