```python
x = 5
y = 5
z = 6

#comparasion operators to build one statement

print(x != y)
print(x != z)
print(x == y) 
print(x == z)
print(x >= y)
print(x > y)
print(x < y)
print(x <= y)

statementA = x != y
print(statementA)

#logical operators to compare multiple statements that each were built with comparasion operators
print(x!=y and x!=z)
print(x!=y or x!=z)
print(not(x!=y))
```

### We can use an if statement to alter the flow of code. After the word "if", we can include an expression that resolves to either true or false. This expression can be built using comparasion and logical operators. If we want two or more thing to happen after our if, we can have an else and elif statement. Ordering matters.

```python
direction = input("left or right")
if direction.strip().lower().startswith("l"):
    print("dragon killed you")
else:
    print("you are alive")
```

```python
number = int(input("Favorite Number: "))

if number >= 10 and number <= 20:
    print("good number")
elif number >= 20:
    print("too large")
else:
    print("too small")
```

```python
clothing_stores = ["walmart", "pandabuy", "nike", "peter millar", "dior"]
fav_store = input("Enter your favorite store: ")
has_store = False

for i in clothing_stores:
    if i == fav_store.lower():
        has_store = True

if has_store == True:
    print("we have that store")
else:
    print("we don't have that store")
```
