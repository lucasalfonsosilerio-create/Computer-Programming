```python
x = 5
y = 5
z = 6

#comparasion operators to build one statement

print(x != y)
print(x != z)
print(x == y) 
print(x == z)
print(x >= y)
print(x > y)
print(x < y)
print(x <= y)

statementA = x != y
print(statementA)

#logical operators to compare multiple statements that each were built with comparasion operators
print(x!=y and x!=z)
print(x!=y or x!=z)
print(not(x!=y))
```
