```python
fav_words = ["poopoo", "peepee", "gaming", "kahoot", "free"]
print(fav_words)
print(fav_words[-1], fav_words[0])
print(fav_words[3].title())

print(f"One of my favorite word is, {fav_words[0]}.")

my_sentence = (f"One of my favorite word is, {fav_words[0]}.")
print(my_sentence)

length = len(fav_words)
print(len(fav_words), length)

print(f"I made a list with the words, {fav_words[-1]} and {fav_words[-3]} and {fav_words[0]}. This list is {length} index long.")
```

# every item in a list starting at 0. we can also start from the end with -1.

```python
birthday = ["april", 29, 2009]
birthday_string = f"I was born in the year {birthday[2]} on the day {birthday[1]}th of the month {birthday[0].title()}."
print(birthday_string)
```

```python
siblings = ["joaquin", "julian", "arianna", "tristan"]
siblings_favor = f"My favorite siblings are, in this order, {siblings[1].title()}, {siblings[2].title()}, {siblings[3].title()}, {siblings[0].title()}."
print(siblings_favor)
siblings.append("charlie")
print(siblings)
siblings[0] = "Charlie"
print(siblings)
```

```python
grades = [23, 98, 12, 130, 87]
grades.insert(0, 67)
print(grades)
print((grades[0] + grades[1] + grades[2] + grades[3] + grades[4])/5)
```
