```python
mysteryNumberGuess = int(input("guess a number"))
guesses = []
guesses.append(mysteryNumberGuess)
print(f"you've taken {len(guesses)} guess")

while mysteryNumberGuess != 4489:
    mysteryNumberGuess = int(input("wrong! guess again"))
    guesses.append(mysteryNumberGuess)
    print(f"you've taken {len(guesses)} guesses")

print(f"you took {len(guesses)} guesses \ncorrect!")

print("your guesses were")
for i in guesses:
    print(i)
```
