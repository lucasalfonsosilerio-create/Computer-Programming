```python
def add_num(a):
    initial = a
    first = 0

    while initial > 0:
        user_num = int(input("what number do you want to add"))
        first = first + user_num

        initial = initial - 1

    print(first)

add_num(4)
```

```python
def summation(x):
    sum = 0
    for i in range(x):
        num = int(input("enter a number"))
        sum = sum + num
    print(sum)

summation(4)
```

```python
def product(x):
    prod = 1
    for i in range(x):
        num = int(input("enter a number: "))
        prod = prod * num
    print(prod)

product(5)
```

```python
def factorial(x):
    prod = 1

    for i in range(1,x+1):
        prod = prod * i

    print(prod)

factorial(7)
```

```python
def manyNum(a):
    initial = a
    skibidi = 0

    while initial > 0:
        num = int(input("insert num: "))
        skibidi = skibidi + num
        initial -= 1

    print(skibidi)

manyNum(5)
```
